```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = 'https://www.jumia.com.ng/mlp-maybelline-new-york-store/'
request = requests.get(url)
#print(request)
soup = BeautifulSoup(request.content, 'html.parser')
#print(soup) 

product = soup.find_all('div', class_ = 'info')
#print(product)
#len(product)

names = []
prices = []
old_prices = []

for i in product:
    if i.find('h3', class_ = 'name') is not None:
        name = i.find('h3', class_ = 'name')
        names.append(name.text)
    else:
        names.append('None')
        
    if i.find('div', class_= 'prc') is not None:
        price = i.find('div', class_= 'prc')
        prices.append(price.text)
    else:
        prices.append('None')
        
    if i.find('div', class_= 'old') is not None:
        old_price = i.find('div', class_= 'old')
        old_prices.append(old_price.text)
    else:
        old_prices.append('None')

#print(names)
#print(prices)
#print(old_prices)

makeup_df = pd.DataFrame({
    'name': names,
    'price': prices,
    'old_price': old_prices
})
print(makeup_df)
```
