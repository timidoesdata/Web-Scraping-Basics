```python
import requests
from bs4 import BeautifulSoup

url = 'https://en.wikipedia.org/wiki/Germany'
response = requests.get(url)
#print(response)
soup = BeautifulSoup(response.content, 'html.parser')
#print(soup)

#1. TO GET PAGE TITLE
title = soup.find('title').text
print(title)

#2. GET A SUB-HEADING
sub = soup.find('div', class_='mw-heading mw-heading3').text
print(sub)

#3. GET ALL SUB-HEADINGS
allsub = soup.find_all('div', class_='mw-heading mw-heading3')
for i in allsub:
    print(i.text)

# 4. GET ALL LINKS
links = soup.find_all('a')
for i in links:
    print(i.get('href'))

#5. CHECK FOR ONE boldened TEXT 
bold = soup.select_one('b').text
print(bold)

#6. CHECK FOR ALL bold TEXT
bold = soup.select('b')
for i in bold:
    print(i.text)
```
